# Data Cleaning and Preparation

This section focuses on validating and preparing
the raw sales dataset for exploratory data analysis.

The cleaning process includes:
- Validating column data types
- Checking unique record identifiers
- Investigating future-dated transactions
- Checking numerical inconsistencies
- Preserving legitimate negative profit values
- Preparing a cleaned dataset for further analysis

```python
# Preserve the original dataset
raw_df = df.copy()

# Create a working copy
clean_df = raw_df.copy()

print("Original dataset:", raw_df.shape)
print("Working dataset:", clean_df.shape)
```

```python
# Convert date columns to datetime
date_columns = ["Order Date", "Ship Date"]

for col in date_columns:
    clean_df[col] = pd.to_datetime(
        clean_df[col],
        errors="coerce"
    )

# Convert numeric columns
numeric_columns = [
    "Row ID",
    "Postal Code",
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

for col in numeric_columns:
    clean_df[col] = pd.to_numeric(
        clean_df[col],
        errors="coerce"
    )

# Verify data types
clean_df.info()
```

```python
print("Duplicate Row IDs:", clean_df["Row ID"].duplicated().sum())

print("Unique Order IDs:", clean_df["Order ID"].nunique())

print("Unique Customers:", clean_df["Customer ID"].nunique())

print("Unique Products:", clean_df["Product ID"].nunique())
```

```python
from datetime import datetime

today = pd.Timestamp.today().normalize()

future_orders = clean_df[
    clean_df["Order Date"] > today
]

future_shipping = clean_df[
    clean_df["Ship Date"] > today
]

print("Future-dated orders:", len(future_orders))
print("Future shipping dates:", len(future_shipping))

display(
    future_orders[
        ["Order ID", "Order Date", "Ship Date", "Sales"]
    ].head(10)
)
```

```python
# Check numerical values
for col in ["Sales", "Quantity", "Discount", "Profit"]:

    print(f"\n--- {col} ---")

    print("Negative values:", (clean_df[col] < 0).sum())
    print("Zero values:", (clean_df[col] == 0).sum())
    print("Missing values:", clean_df[col].isna().sum())

# Check discount range
print("\nDiscount outside 0-1 range:")

print(
    (
        (clean_df["Discount"] < 0) |
        (clean_df["Discount"] > 1)
    ).sum()
)
```

```python
invalid_shipping_dates = clean_df[
    clean_df["Ship Date"] < clean_df["Order Date"]
]

print(
    "Shipping dates before order dates:",
    len(invalid_shipping_dates)
)
```

```python
def detect_outliers(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    print(f"\n{column} Outlier Analysis")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Potential Outliers:", len(outliers))

    return outliers


sales_outliers = detect_outliers(clean_df, "Sales")
profit_outliers = detect_outliers(clean_df, "Profit")

display(sales_outliers.head(10))
display(profit_outliers.head(10))
```

```python
total_orders = clean_df["Order ID"].nunique()
```

```python
# Fixed analysis cutoff date
analysis_cutoff = pd.Timestamp("2026-11-16")

# Identify orders placed after the cutoff
future_order_records = clean_df[
    clean_df["Order Date"] > analysis_cutoff
]

# Keep only orders placed on or before the cutoff
analysis_df = clean_df[
    clean_df["Order Date"] <= analysis_cutoff
].copy()

# Identify retained orders that were shipped after the cutoff
pending_shipping = analysis_df[
    analysis_df["Ship Date"] > analysis_cutoff
]

print("Original records:", len(clean_df))
print("Future order records excluded:", len(future_order_records))
print("Historical records retained:", len(analysis_df))
print("Retained orders with future shipping dates:", len(pending_shipping))
print("Unique future orders excluded:",
      future_order_records["Order ID"].nunique())
```

```python
# Final quality checks
print("Dataset shape:", analysis_df.shape)

print("Missing values:", analysis_df.isnull().sum().sum())

print("Duplicate rows:", analysis_df.duplicated().sum())

print("Duplicate Row IDs:",
      analysis_df["Row ID"].duplicated().sum())

print("Date range:",
      analysis_df["Order Date"].min(),
      "to",
      analysis_df["Order Date"].max())

print("Total unique orders:",
      analysis_df["Order ID"].nunique())
```

```python
# Save processed data
analysis_df.to_csv(
    "../data/processed_sales_data.csv",
    index=False
)

print("Processed dataset saved successfully!")
```

## Data Cleaning Summary

The dataset underwent initial data quality assessment and preparation.

### Findings
- No missing values were identified.
- No exact duplicate records were identified.
- All Row IDs were unique.
- Negative and zero profit values were retained.
- Zero discount values were treated as valid.
- No invalid discount ranges were identified.
- No shipping dates occurred before order dates.

### Date Handling
A fixed analysis cutoff of November 16, 2026 was established.

Orders placed after this date were excluded from historical sales analysis.

Orders placed before or on the cutoff date were retained, even if their shipping dates were later than the cutoff.

### Data Preservation
The original dataset was preserved.
The processed historical dataset was exported as a separate CSV file.

# Exploratory Data Analysis (EDA)

## 1. Overall Business Performance

This section evaluates the overall sales performance of the business using key metrics such as revenue, profit, orders, quantity sold, customers and average order value.

```python
# Overall business metrics

total_revenue = analysis_df["Sales"].sum()

total_profit = analysis_df["Profit"].sum()

total_orders = analysis_df["Order ID"].nunique()

total_quantity = analysis_df["Quantity"].sum()

total_customers = analysis_df["Customer ID"].nunique()

total_products = analysis_df["Product ID"].nunique()

# Calculate revenue for each individual order
order_revenue = (
    analysis_df.groupby("Order ID")["Sales"].sum()
)

# Average Order Value
average_order_value = order_revenue.mean()

# Overall profit margin
profit_margin = (total_profit / total_revenue) * 100

# Average recorded discount
average_discount = analysis_df["Discount"].mean() * 100

# Display results
print("========== OVERALL BUSINESS PERFORMANCE ==========")

print(f"Total Revenue: {total_revenue:,.2f}")
print(f"Total Profit: {total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Unique Products: {total_products:,}")
print(f"Average Order Value: {average_order_value:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Average Recorded Discount: {average_discount:.2f}%")
```

```python
category_performance = (
    analysis_df.groupby("Category")
    .agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum"),
        Unique_Orders=("Order ID", "nunique")
    )
    .sort_values("Total_Revenue", ascending=False)
)

display(category_performance.round(2))
```

## 2. Sales and Revenue Trends

This section examines monthly and yearly sales patterns to identify revenue growth, fluctuations and seasonal trends.

```python
# Group sales by month

monthly_sales = (
    analysis_df.groupby(
        analysis_df["Order Date"].dt.to_period("M")
    )
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
    .reset_index()
)

monthly_sales["Order Date"] = (
    monthly_sales["Order Date"].dt.to_timestamp()
)

display(monthly_sales.head(12))
```

```python
import plotly.express as px

fig = px.line(
    monthly_sales,
    x="Order Date",
    y="Revenue",
    title="Monthly Revenue Trend",
    markers=True,
    labels={
        "Order Date": "Month",
        "Revenue": "Revenue"
    }
)

fig.update_layout(
    template="plotly_white",
    hovermode="x unified",
    xaxis_title="Month",
    yaxis_title="Revenue"
)

fig.show()
```

```python
yearly_sales = (
    analysis_df.groupby(
        analysis_df["Order Date"].dt.year
    )
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .reset_index()
)

yearly_sales.rename(
    columns={"Order Date": "Year"},
    inplace=True
)

display(yearly_sales.round(2))
```

```python
fig = px.bar(
    yearly_sales,
    x="Year",
    y="Revenue",
    title="Annual Revenue Comparison",
    text_auto=".2s",
    labels={
        "Revenue": "Total Revenue",
        "Year": "Year"
    }
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Year",
    yaxis_title="Revenue"
)

fig.show()
```

```python
#Performance by sub-category

subcategory_performance = (
    analysis_df.groupby(["Category", "Sub-Category"])
    .agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum"),
        Unique_Orders=("Order ID", "nunique")
    )
    .reset_index()
)

subcategory_performance["Profit_Margin"] = (
    subcategory_performance["Total_Profit"]
    / subcategory_performance["Total_Revenue"] * 100
)

subcategory_performance = subcategory_performance.sort_values(
    "Total_Revenue", ascending=False
)

display(subcategory_performance.round(2))
```

```python
#Top 10 products by revenue

top_products = (
    analysis_df.groupby("Product Name")
    .agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
    .sort_values("Total_Revenue", ascending=False)
    .head(10)
)

display(top_products.round(2))
```

```python
#Top 10 products by profit

top_profit_products = (
    analysis_df.groupby("Product Name")
    .agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum")
    )
    .sort_values("Total_Profit", ascending=False)
    .head(10)
)

display(top_profit_products.round(2))
```

```python
#Products generating losses

loss_products = (
    analysis_df.groupby("Product Name")
    .agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Units_Sold=("Quantity", "sum")
    )
    .query("Total_Profit < 0")
    .sort_values("Total_Profit")
)

display(loss_products.head(10).round(2))
```

```python
#Performance by region

region_performance = (
    analysis_df.groupby("Region")
    .agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Customers=("Customer ID", "nunique"),
        Units_Sold=("Quantity", "sum")
    )
)

region_performance["Profit_Margin"] = (
    region_performance["Total_Profit"]
    / region_performance["Total_Revenue"] * 100
)

display(region_performance.sort_values(
    "Total_Revenue", ascending=False
).round(2))
```

```python
#Performance by state or province

state_performance = (
    analysis_df.groupby("State/Province")
    .agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Customers=("Customer ID", "nunique")
    )
)

state_performance["Profit_Margin"] = (
    state_performance["Total_Profit"]
    / state_performance["Total_Revenue"] * 100
)

display(state_performance.sort_values(
    "Total_Revenue", ascending=False
).head(10).round(2))
```

```python
#Visualize revenue by region

import plotly.express as px

region_chart_data = (
    region_performance
    .reset_index()
    .sort_values("Total_Revenue", ascending=False)
)

fig = px.bar(
    region_chart_data,
    x="Region",
    y="Total_Revenue",
    color="Region",
    title="Revenue by Region",
    text_auto=".2s"
)

fig.update_layout(
    template="plotly_white",
    showlegend=False,
    xaxis_title="Region",
    yaxis_title="Revenue"
)

fig.show()
```

#Customer Performance Analysis

```python
#Customer-level performance

customer_performance = (
    analysis_df.groupby(["Customer ID", "Customer Name"])
    .agg(
        Total_Revenue=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Orders=("Order ID", "nunique"),
        Total_Quantity=("Quantity", "sum")
    )
)

customer_performance["Average_Order_Value"] = (
    customer_performance["Total_Revenue"]
    / customer_performance["Total_Orders"]
)

display(
    customer_performance
    .sort_values("Total_Revenue", ascending=False)
    .head(10)
    .round(2)
)
```

```python
#Top 10 customers by revenue

top_customers = (
    customer_performance
    .sort_values("Total_Revenue", ascending=False)
    .head(10)
)

fig = px.bar(
    top_customers.reset_index(),
    x="Total_Revenue",
    y="Customer Name",
    orientation="h",
    title="Top 10 Customers by Revenue",
    text_auto=".2s"
)

fig.update_layout(
    template="plotly_white",
    yaxis={"categoryorder": "total ascending"},
    xaxis_title="Revenue",
    yaxis_title="Customer"
)

fig.show()
```

```python
#Repeat vs one-time customers

customer_order_summary = (
    analysis_df.groupby("Customer ID")["Order ID"]
    .nunique()
)

repeat_customers = (customer_order_summary > 1).sum()
one_time_customers = (customer_order_summary == 1).sum()

repeat_customer_percentage = (
    repeat_customers / len(customer_order_summary) * 100
)

print("Total Customers:", len(customer_order_summary))
print("Repeat Customers:", repeat_customers)
print("One-Time Customers:", one_time_customers)
print(f"Repeat Customer Rate: {repeat_customer_percentage:.2f}%")
```
